## Obtain geographical coordinates for pws names

### Subtask:
Find the latitude and longitude for each unique PWS Name.


**Reasoning**:
Import the Nominatim geocoder, instantiate it, and then iterate through the unique PWS names to get their latitude and longitude, creating a DataFrame with the results and reporting the number of names not found.

## Obtain geographical coordinates for zip codes per state (using pgeocode)

### Subtask:
Find the latitude and longitude for unique zip codes within each state using the `pgeocode` library.

```python
import pgeocode
import pandas as pd

# Initialize the Nominatim geocoder for the US using pgeocode
nomi = pgeocode.Nominatim('us')

# Get all unique states from the combined dataset
unique_states = UCMR5_data_combined['State'].unique()

# Create a dictionary to store zip code locations for each state
all_states_zip_locations = {}

print("Starting geocoding process for all states using pgeocode...")

# Iterate through each unique state
for state in unique_states:
    print(f"\nProcessing state: {state}")

    # Filter the DataFrame for rows where the State is the current state
    state_data = UCMR5_data_combined[UCMR5_data_combined['State'] == state].copy()

    # Get the unique zip codes for the current state
    unique_state_zip_codes = state_data['ZIPCODE'].unique()

    state_latitudes = []
    state_longitudes = []
    state_zip_codes_list = []

    print(f"Attempting to geocode {len(unique_state_zip_codes)} unique zip codes for {state}...")

    # Use pgeocode to get location info for each zip code in the current state
    # pgeocode can query multiple postal codes at once, which might be more efficient
    zip_info_df = nomi.query_postal_code(unique_state_zip_codes.tolist())

    # Process the results
    for zipcode_code in unique_state_zip_codes:
        zip_info = zip_info_df[zip_info_df.postal_code == str(zipcode_code)].iloc[0] # Get the row for the current zipcode

        if not pd.isna(zip_info['latitude']) and not pd.isna(zip_info['longitude']):
            state_latitudes.append(zip_info['latitude'])
            state_longitudes.append(zip_info['longitude'])
            state_zip_codes_list.append(zipcode_code)
        else:
            state_latitudes.append(None)
            state_longitudes.append(None)
            state_zip_codes_list.append(zipcode_code)
            print(f"Could not find coordinates for {state} zip code using pgeocode: {zipcode_code}")


    # Create a new DataFrame with unique zip codes and their coordinates for the current state
    state_zip_locations_pgeocode = pd.DataFrame({
        'ZIPCODE': state_zip_codes_list,
        'Latitude': state_latitudes,
        'Longitude': state_longitudes
    })

    # Store the results in the dictionary
    all_states_zip_locations[state] = state_zip_locations_pgeocode

    # Check how many zip codes were not found for the current state
    not_found_count_pgeocode = state_zip_locations_pgeocode['Latitude'].isnull().sum()
    print(f"Number of {state} zip codes not found using pgeocode: {not_found_count_pgeocode}")

print("\nGeocoding process for all states complete.")

# Optionally, display the head of the DataFrames for a few states
print("\nZip Code Locations per State (Examples using pgeocode):")
for state in list(all_states_zip_locations.keys())[:3]: # Display first 3 states
    print(f"\n--- {state} ---")
    display(all_states_zip_locations[state].head())
```

```python
import pgeocode
import pandas as pd

# Initialize the Nominatim geocoder for the US using pgeocode
nomi = pgeocode.Nominatim('us')

# Filter the missing zip codes to identify the 4-digit ones
four_digit_missing_zips = [zipcode for zipcode in missing_zip_codes if len(str(zipcode)) == 4]

# Prepend '0' to the 4-digit zip codes
corrected_four_digit_zips = ['0' + str(zipcode) for zipcode in four_digit_missing_zips]

# Identify any missing zip codes that are already 5 digits (or have a different number of digits)
other_missing_zips = [zipcode for zipcode in missing_zip_codes if len(str(zipcode)) != 4]

# Combine the corrected 4-digit zips with the other missing zips
all_missing_zips_corrected = corrected_four_digit_zips + other_missing_zips

print(f"Attempting to re-geocode {len(all_missing_zips_corrected)} missing zip codes after correcting 4-digit ones...")

# Use pgeocode to query the updated list of missing zip codes
re_geocoded_missing_zips_df = nomi.query_postal_code(all_missing_zips_corrected)

# Create a DataFrame to store the re-geocoded locations
re_geocoded_locations = pd.DataFrame({
    'ZIPCODE': re_geocoded_missing_zips_df['postal_code'].tolist(),
    'Latitude': re_geocoded_missing_zips_df['latitude'].tolist(),
    'Longitude': re_geocoded_missing_zips_df['longitude'].tolist()
})

print("\nRe-geocoded Locations for Missing Zip Codes:")
display(re_geocoded_locations.head())

# Check how many were found in this re-geocoding step
re_found_count = re_geocoded_locations['Latitude'].notnull().sum()
print(f"\nNumber of previously missing zip codes successfully re-geocoded: {re_found_count}")

re_not_found_count = re_geocoded_locations['Latitude'].isnull().sum()
print(f"Number of previously missing zip codes still not found: {re_not_found_count}")

# Update the all_states_zip_locations dictionary with the newly found coordinates
# Iterate through the states and update the corresponding zip codes
for index, row in re_geocoded_locations.iterrows():
    zipcode_code = row['ZIPCODE']
    latitude = row['Latitude']
    longitude = row['Longitude']

    if pd.notnull(latitude) and pd.notnull(longitude):
        # Find which state this zipcode belongs to in the original data to update the correct state DataFrame
        original_row = UCMR5_data_combined[UCMR5_data_combined['ZIPCODE'] == zipcode_code]
        if not original_row.empty:
            state = original_row['State'].iloc[0]
            if state in all_states_zip_locations:
                # Find the row for this zipcode in the state's location DataFrame
                state_df = all_states_zip_locations[state]
                # Ensure the ZIPCODE column is consistent type for comparison
                state_df['ZIPCODE'] = state_df['ZIPCODE'].astype(str)
                zipcode_code = str(zipcode_code) # Ensure the lookup value is also string

                if zipcode_code in state_df['ZIPCODE'].values:
                    # Update the Latitude and Longitude for this zipcode
                    state_df.loc[state_df['ZIPCODE'] == zipcode_code, 'Latitude'] = latitude
                    state_df.loc[state_df['ZIPCODE'] == zipcode_code, 'Longitude'] = longitude
                else:
                     # This case shouldn't happen if missing_zip_codes was correctly derived
                     print(f"Warning: Re-geocoded zipcode {zipcode_code} not found in {state}'s location DataFrame.")


print("\nUpdated all_states_zip_locations dictionary with re-geocoded coordinates.")
```

```python
# Examine the mapping_data DataFrame
print("Mapping Data DataFrame columns and types:")
mapping_data.info()

# Inspect unique contaminants to identify those of interest
print("\nUnique Contaminants (first 20):")
display(mapping_data['Contaminant'].unique()[:20])

# Inspect unique values in 'Result (µg/L)' to understand how to convert to numeric
print("\nUnique values in 'Result (µg/L)' (first 20):")
display(mapping_data['Result (µg/L)'].unique()[:20])

# Convert 'Result (µg/L)' to numeric, coercing errors to NaN
# Replace '< MRL' and other non-numeric values with NaN
mapping_data['Result_numeric'] = pd.to_numeric(mapping_data['Result (µg/L)'], errors='coerce')

# Define contaminants of interest (example: assuming PFAS substances are of interest)
# Based on the unique contaminants observed, let's select a few common PFAS.
# In a real scenario, this would be based on task requirements or regulatory lists.
contaminants_of_interest = ['PFOA', 'PFOS', 'PFNA', 'PFHxS', 'PFHpA']

# Filter the DataFrame for the contaminants of interest
filtered_contaminants_df = mapping_data[mapping_data['Contaminant'].isin(contaminants_of_interest)].copy()

# Define a concentration threshold (example: above reporting limit or a specific health advisory level)
# For this example, let's consider results above 0 as potentially relevant for a hotspot,
# ignoring '< MRL' which were converted to NaN. We can adjust this threshold based on regulations if known.
concentration_threshold = 0

# Filter for results above the threshold, excluding NaNs in 'Result_numeric'
hotspot_data = filtered_contaminants_df[filtered_contaminants_df['Result_numeric'] > concentration_threshold].copy()

# Display the first few rows and information of the hotspot_data DataFrame
print("\nHotspot Data DataFrame Head (Filtered):")
display(hotspot_data.head())

print("\nHotspot Data DataFrame Info (Filtered):")
hotspot_data.info()
```

**Reasoning**:
Group the hotspot data by geographical location and contaminant to aggregate results for mapping.



```python
# Group by Latitude, Longitude, and Contaminant to aggregate data for mapping
# For each group, we can aggregate relevant information, e.g., count occurrences,
# calculate the mean/max concentration, and list the PWS Names/IDs.

# Since we want to show all contaminants at a location, let's group by Latitude and Longitude,
# and then aggregate the contaminant information.
# We will also keep PWSID and PWS Name to include in the popup information.

# Group by Latitude and Longitude, and aggregate relevant columns
# Using a dictionary for aggregation: list unique contaminants, PWSIDs, and PWS Names
grouped_hotspot_data = hotspot_data.groupby(['Latitude', 'Longitude']).agg(
    Contaminants=('Contaminant', lambda x: list(x.unique())),
    PWSIDs=('PWSID', lambda x: list(x.unique())),
    PWSNames=('PWS Name', lambda x: list(x.unique())),
    Max_Result=('Result_numeric', 'max'),
    Count=('Contaminant', 'count') # Count the number of occurrences at this location
).reset_index()

# Convert lists in aggregated columns to a more readable string format for popups
grouped_hotspot_data['Contaminants_List'] = grouped_hotspot_data['Contaminants'].apply(lambda x: ', '.join(x))
grouped_hotspot_data['PWSIDs_List'] = grouped_hotspot_data['PWSIDs'].apply(lambda x: ', '.join(x))
grouped_hotspot_data['PWSNames_List'] = grouped_hotspot_data['PWSNames'].apply(lambda x: ', '.join(x))


# Display the first few rows of the grouped data
print("\nGrouped Hotspot Data (aggregated by location):")
display(grouped_hotspot_data.head())

# Display info of the grouped data
print("\nGrouped Hotspot Data Info:")
grouped_hotspot_data.info()
```

**Reasoning**:
Generate a map using the grouped hotspot data, adding markers with popups that display the aggregated contaminant information, PWS IDs, PWS Names, and the maximum result at each location.



```python
import folium

# Calculate the approximate center of the grouped hotspot data
center_lat_hotspot = grouped_hotspot_data['Latitude'].mean()
center_lon_hotspot = grouped_hotspot_data['Longitude'].mean()

# Create a base map centered on the hotspot data
m_hotspot = folium.Map(location=[center_lat_hotspot, center_lon_hotspot], zoom_start=7)

# Add markers for each hotspot location
for index, row in grouped_hotspot_data.iterrows():
    # Create a popup with aggregated information
    popup_text = f"Latitude: {row['Latitude']:.4f}<br>" \
                 f"Longitude: {row['Longitude']:.4f}<br>" \
                 f"Contaminants: {row['Contaminants_List']}<br>" \
                 f"PWS Names: {row['PWSNames_List']}<br>" \
                 f"Max Result (µg/L): {row['Max_Result']:.4f}<br>" \
                 f"Number of Detections: {row['Count']}"

    # Add a marker to the map
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=popup_text,
        icon=folium.Icon(color='red', icon='info-sign') # Using a different color/icon for hotspots
    ).add_to(m_hotspot)

print("Hotspot markers added to the map.")

# Display the map
m_hotspot
```

I will iterate through a representative point (like the closest point on the geometry to the hotspot) for each river geometry instead of just the centroid, and use geodesic distance.
```python
from geopy.distance import geodesic
from shapely.geometry import Point, MultiPolygon, Polygon, MultiLineString, LineString
import numpy as np
import pandas as pd

def calculate_min_distance_to_river_geometry(row, river_gdf):
    """
    Calculates the minimum geodesic distance from a hotspot location to the
    geometries in the Delaware River GeoDataFrame by finding the distance
    to representative points on the river geometries.

    Args:
        row: A row from the grouped_hotspot_data DataFrame containing Latitude
             and Longitude of the hotspot.
        river_gdf: The GeoDataFrame containing the Delaware River geometries.

    Returns:
        The minimum distance in miles as a float, or np.nan if coordinates
        are missing or an error occurs.
    """
    hotspot_coords = (row['Latitude'], row['Longitude'])

    # Check if hotspot coordinates are valid
    if pd.isna(hotspot_coords[0]) or pd.isna(hotspot_coords[1]):
        return np.nan

    min_distance = np.inf
    hotspot_point = Point(hotspot_coords[1], hotspot_coords[0]) # Shapely expects (lon, lat)

    # Iterate through each geometry in the river GeoDataFrame
    for river_geometry in river_gdf.geometry:
        try:
            if river_geometry.is_empty:
                continue

            # Get a representative point on the river geometry that is closest to the hotspot
            # This is more robust than using the centroid for all geometry types
            closest_point_on_river = river_geometry.interpolate(river_geometry.project(hotspot_point))

            # Get the coordinates of the closest point (longitude, latitude)
            closest_coords = (closest_point_on_river.y, closest_point_on_river.x)

            # Calculate the geodesic distance in miles
            distance = geodesic(hotspot_coords, closest_coords).miles
            min_distance = min(min_distance, distance)

        except Exception as e:
            # Print a warning but continue with other geometries
            print(f"Warning: Error calculating distance for a river geometry: {e}")
            pass # Suppress frequent warnings


    # If min_distance is still infinity, it means no valid river geometries were processed
    if min_distance == np.inf:
         # Fallback: if GeoDataFrame processing failed or resulted in no valid distances,
         # calculate distance to the approximated river points instead.
         print(f"Warning: Could not calculate distance to GeoDataFrame geometry for hotspot at {hotspot_coords}. Falling back to approximated points.")
         min_distance_approx = np.inf
         for index, river_row in delaware_river_approx_coords.iterrows():
             river_coords = (river_row['Latitude'], river_row['Longitude'])
             if pd.isna(river_coords[0]) or pd.isna(river_coords[1]):
                 continue
             try:
                 distance_approx = geodesic(hotspot_coords, river_coords).miles
                 min_distance_approx = min(min_distance_approx, distance_approx)
             except Exception as e:
                 print(f"Error calculating fallback distance for hotspot {hotspot_coords} to river point {river_coords}: {e}")
                 pass
         return min_distance_approx if min_distance_approx != np.inf else np.nan

    else:
        return min_distance


# Apply the function to each row of the grouped_hotspot_data DataFrame,
# passing the loaded delaware_river_gdf
grouped_hotspot_data['Distance_to_River_miles'] = grouped_hotspot_data.apply(
    calculate_min_distance_to_river_geometry, axis=1, river_gdf=delaware_river_gdf
)

# Display the first few rows of the DataFrame with the new column
print("\nGrouped Hotspot Data with Distance to River (using GeoDataFrame with refined geometry handling):")
display(grouped_hotspot_data.head())
```

## Obtain geographical coordinates for zip codes

### Subtask:
Find the latitude and longitude for each unique zip code.

**Reasoning**:
Use the `pgeocode` library to find the latitude and longitude for each unique zip code and create a DataFrame with the results.

```python
import pgeocode
import pandas as pd

# Ensure unique_zip_codes is defined by extracting from merged_df
# Initialize an empty list to store all zip codes
all_zip_codes = []

# Iterate through the 'ZIPCODES' column
for zip_list in merged_df['ZIPCODES']:
    # Check if the entry is a list and not None or NaN
    if isinstance(zip_list, list):
        # Extend the all_zip_codes list with the zip codes from the current list
        all_zip_codes.extend(zip_list)
    elif pd.notnull(zip_list):
        # If it's not a list but not NaN (shouldn't happen based on aggregation, but good practice)
        # Append the single zip code. Ensure it's treated as a string.
        all_zip_codes.append(str(zip_list))


# Get the unique zip codes from the collected list
unique_zip_codes = pd.Series(all_zip_codes).unique()

print("Unique Zip Codes extracted from merged_df:")
display(unique_zip_codes[:10]) # Displaying the first 10 for brevity
print(f"\nTotal number of unique zip codes: {len(unique_zip_codes)}")


# Initialize the Nominatim geocoder for the US using pgeocode
nomi = pgeocode.Nominatim('us')

latitudes = []
longitudes = []
zip_codes_list = []

print(f"\nAttempting to geocode {len(unique_zip_codes)} unique zip codes using pgeocode...")

# Prepare zip codes for query: add leading zero to 4-digit zip codes and convert to strings
zip_codes_to_query = []
for code in unique_zip_codes:
    code_str = str(code)
    if len(code_str) == 4:
        zip_codes_to_query.append('0' + code_str)
    else:
        zip_codes_to_query.append(code_str)


# Query pgeocode for all unique zip codes
zip_info_df = nomi.query_postal_code(zip_codes_to_query)

# Process the results
for index, row in zip_info_df.iterrows():
    zipcode_code = row['postal_code']
    latitude = row['latitude']
    longitude = row['longitude']

    zip_codes_list.append(zipcode_code)

    if not pd.isna(latitude) and not pd.isna(longitude):
        latitudes.append(latitude)
        longitudes.append(longitude)
    else:
        latitudes.append(None)
        longitudes.append(None)
        print(f"Could not find coordinates for zip code using pgeocode: {zipcode_code}")


# Create a new DataFrame with unique zip codes and their coordinates
zip_code_locations = pd.DataFrame({
    'ZIPCODE': zip_codes_list,
    'Latitude': latitudes,
    'Longitude': longitudes
})

print("\nZip Code Locations DataFrame (using pgeocode):")
display(zip_code_locations.head())

# Check how many zip codes were not found
not_found_count = zip_code_locations['Latitude'].isnull().sum()
print(f"\nNumber of zip codes not found using pgeocode: {not_found_count}")
```

**Reasoning**:
Define a function to correct 4-digit zip codes by adding a leading '0' and apply this function to the 'ZIPCODES' column of merged_df.



```python
def correct_zipcodes(zip_list):
    """
    Corrects 4-digit zip codes in a list by adding a leading '0'.

    Args:
        zip_list: A list of zip codes (can contain integers or strings).

    Returns:
        A new list with corrected zip code strings.
    """
    corrected_list = []
    if isinstance(zip_list, list):
        for zipcode in zip_list:
            zipcode_str = str(zipcode)
            if len(zipcode_str) == 4:
                corrected_list.append('0' + zipcode_str)
            else:
                corrected_list.append(zipcode_str)
    # Handle cases where ZIPCODES might not be a list (e.g., NaN after left merge)
    elif pd.notnull(zip_list):
         zipcode_str = str(zip_list)
         if len(zipcode_str) == 4:
             corrected_list.append('0' + zipcode_str)
         else:
             corrected_list.append(zipcode_str)

    return corrected_list

# Apply the correct_zipcodes function to the 'ZIPCODES' column
merged_df['ZIPCODES'] = merged_df['ZIPCODES'].apply(correct_zipcodes)

print("Merged DataFrame with corrected ZIPCODES:")
display(merged_df.head())
```
